```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
adults = df[(df['age'] >= 18) & (df['fare'] > 30)]
adults.head
```

```python
df['family_size'] = df['sibsp'] + df['parch'] + 1
df[['sibsp' , 'parch', 'family_size']].head()
```

```python
df['fare_per_person'] = df['fare'] / df['family_size']
```

```python
df[['fare', 'family_size', 'fare_per_person']].head()
```

```python
df.columns
```

```python
df.groupby(['pclass' , 'sex'])['fare'].mean()
```

```python
df.groupby(['pclass','sex']) ['survived'].mean()
```

```python
analysis = df.groupby(['pclass','sex']).agg({
    'fare': 'mean',
    'survived':'mean'
})
analysis
```
